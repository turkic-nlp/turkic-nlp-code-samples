# Khalaj (klj) — Tokenisation and Translation

Khalaj (Arghu branch, Arabic script) currently has rule-based tokenisation and prototype-quality morphological analysis. NLLB-200 provides cross-lingual embeddings and machine translation. Khalaj belongs to the Arghu branch, the most divergent Turkic branch.

```python
import turkicnlp
from turkicnlp import Pipeline
```

## 1. Tokenisation

```python
turkicnlp.download("klj")
nlp_tok = Pipeline("klj", processors=["tokenize"])
doc = nlp_tok("من مکتبه گئدیرم.")
print([w.text for w in doc.words])
```

## 2. Morphological Analysis (Apertium FST — Prototype)

```python
nlp = Pipeline(
    "klj",
    processors=["tokenize", "morph"],
    morph_backend="apertium",
)
doc = nlp("من مکتبه گئدیرم.")
for w in doc.words:
    print(f"{w.text:<18} lemma={w.lemma} feats={w.feats}")
```

## 3. Translation via NLLB-200

```python
turkicnlp.download("klj", processors=["translate"])
trans = Pipeline("klj", processors=["translate"], translate_tgt_lang="eng_Latn")
doc = trans("من مکتبه گئدیرم.")
print("EN:", doc.translation)
```
