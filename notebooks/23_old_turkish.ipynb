# Old Turkish (otk) — Tokenisation and Translation

Old Turkish (Historical branch, Old Turkic Runic script) currently has rule-based tokenisation and prototype-quality morphological analysis. NLLB-200 provides cross-lingual embeddings and machine translation. Old Turkish is used for Orkhon and Yenisei runic inscriptions.

```python
import turkicnlp
from turkicnlp import Pipeline
```

## 1. Tokenisation

```python
from turkicnlp.scripts import Script
from turkicnlp.scripts.detector import detect_script
from turkicnlp.scripts.transliterator import Transliterator

# Old Turkish Runic text (Orkhon inscriptions)
runic = "𐰼𐰀𐰭 𐰕𐰇𐰤𐱅𐰞𐰺𐰀"
print("Script Detection:")
print(f"  Detected: {detect_script(runic).name}")
print()

# Old Turkic Runic -> Turkic Common Alphabet (Latin)
try:
    t = Transliterator("otk", source=Script.OLD_TURKIC, target=Script.COMMON_TURKIC)
    common = t.transliterate(runic)
    print(f"Old Turkic Runic: {runic}")
    print(f"Turkic Common:    {common}")
    print("\n✓ Old Turkish transliteration to Turkic Common Alphabet (Latin) successful")
except Exception as e:
    print(f"⚠ Note: Transliteration to Turkic Common Alphabet (Latin) may not be fully supported for Old Turkish: {e}")
    print("  Transliteration of historical scripts requires specialized tools")
```

## 2. Script Detection and Old Turkic Runic Transliteration

Old Turkish is written in Old Turkic Runic script (also called Orkhon/Yenisei script). The transliteration system enables conversion to Latin for processing historical texts.

```python
turkicnlp.download("otk")
nlp_tok = Pipeline("otk", processors=["tokenize"])
doc = nlp_tok("𐰼𐰀𐰭 𐰕𐰇𐰤𐱅𐰞𐰺𐰀")
print([w.text for w in doc.words])
```

## 2. Morphological Analysis (Apertium FST — Prototype)

```python
nlp = Pipeline(
    "otk",
    processors=["tokenize", "morph"],
    morph_backend="apertium",
)
doc = nlp("𐰼𐰀𐰭 𐰕𐰇𐰤𐱅𐰞𐰺𐰀")
for w in doc.words:
    print(f"{w.text:<18} lemma={w.lemma} feats={w.feats}")
```

## 3. Translation via NLLB-200

```python
turkicnlp.download("otk", processors=["translate"])
trans = Pipeline("otk", processors=["translate"], translate_tgt_lang="eng_Latn")
doc = trans("𐰼𐰀𐰭 𐰕𐰇𐰤𐱅𐰞𐰺𐰀")
print("EN:", doc.translation)
```
