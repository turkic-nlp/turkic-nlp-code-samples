# Gagauz (gag) — Tokenisation and Morphological Analysis

Gagauz (Oghuz branch) is supported via Latin script tokenisation and Prototype-quality Apertium FST morphological analysis. NLLB-200 provides cross-lingual embeddings and machine translation. Gagauz is spoken primarily in Moldova.

```python
# Install TurkicNLP
# pip install turkicnlp          # core (tokenization, transliteration)
# pip install "turkicnlp[stanza]"  # adds POS, lemma, depparse, NER
# pip install "turkicnlp[nllb]"    # adds cross-lingual embeddings + translation
# pip install "turkicnlp[all]"     # all optional dependencies
```

```python
import turkicnlp
from turkicnlp import Pipeline
```

## 1. Download Models

```python
turkicnlp.download('gag')
```

## 2. Tokenisation

```python
nlp_tok = Pipeline("gag", processors=["tokenize"])
doc = nlp_tok("Ben mektebi gidiyom.")
print([w.text for w in doc.words])
```

## 3. Morphological Analysis (Apertium FST — Prototype)

```python
nlp = Pipeline(
    "gag",
    processors=["tokenize", "morph"],
    morph_backend="apertium",
)
doc = nlp("Ben mektebi gidiyom.")
for w in doc.words:
    print(f"{w.text:<18} lemma={w.lemma:<12} feats={w.feats}")
```

## 4. Translation via NLLB-200

```python
turkicnlp.download("gag", processors=["translate"])
trans = Pipeline("gag", processors=["translate"], translate_tgt_lang="eng_Latn")
doc = trans("Ben mektebi gidiyom.")
print("EN:", doc.translation)
```
